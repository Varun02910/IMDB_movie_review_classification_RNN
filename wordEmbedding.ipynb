```python
from tensorflow.keras.preprocessing.text import one_hot

import tensorflow as tf
one_hot = tf.keras.preprocessing.text.one_hot
```

```python
sents = ['the glass of milk', 
          'the glass of juice',
          'the cup of tea',
          'I am an amazing boy',
          'I  am a good developer',
          'Understand the meaning of words',
          'your video is awesome']
```

```python
##define the vocabulary size
voc_size=10000
```

```python
## One hot representation
one_hot_repre = [one_hot(sent, voc_size) for sent in sents]
one_hot_repre
```

```python
#word embedding representation
## importing important libraries/classes
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
## right now the sentences are of different length, so we need to make them of same length, for htat we will use the pas_sequences

sent_length = 8
embedded_docs = pad_sequences(one_hot_repre, padding='pre', maxlen=sent_length)
embedded_docs
```

```python
dim = 10
```

```python
## feature representation using word embedding

model = Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```
